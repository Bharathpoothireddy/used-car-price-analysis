```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
print("Libraries imported successfully")
```

```python
df = pd.read_csv(r"used_car_dataset.csv")
print(" Dataset loaded successfully")
```

```python
sns.set_style('darkgrid')
plt.rcParams['figure.figsize'] = (12, 6)
```

```python
print("First 5 rows:")
df.head()
```

```python
print("Data Types:")
df.info()
```

```python
df['AskPrice'] = df['AskPrice'].astype(str).str.replace(r'[₹,\s]', '', regex=True)
df['AskPrice'] = pd.to_numeric(df['AskPrice'], errors='coerce')
```

```python
df['kmDriven'] = df['kmDriven'].astype(str).str.replace(r'[,]', '', regex=True)  # Remove commas first
df['kmDriven'] = df['kmDriven'].str.replace(r'[kKmM]', '', regex=True)  # Remove k or m (case insensitive)
df['kmDriven'] = df['kmDriven'].str.strip()  # Remove any extra spaces
df['kmDriven'] = pd.to_numeric(df['kmDriven'], errors='coerce')
```

```python
print(f"✅ AskPrice converted to {df['AskPrice'].dtype}")
print(f"   Min: {df['AskPrice'].min():,.0f}, Max: {df['AskPrice'].max():,.0f}")
print(f"   Rows with invalid prices: {df['AskPrice'].isnull().sum()}\n")

# Convert kmDriven - Remove km, commas, and spaces
print(f"✅ KmDriven converted to {df['kmDriven'].dtype}")
print(f"   Min: {df['kmDriven'].min():,.0f}, Max: {df['kmDriven'].max():,.0f}")
print(f"   Rows with invalid prices: {df['kmDriven'].isnull().sum()}\n")
```

```python
print(f"\n📊 Dataset Size: {df.shape[0]:,} rows, {df.shape[1]} columns")

print("\n📋 Column Details:")
for col in df.columns:
    dtype = df[col].dtype.name
    missing = df[col].isnull().sum()
    unique = df[col].nunique()
    print(f"  {col:20s} | Type: {dtype:12s} | Missing: {missing:5d} | Unique: {unique}")
```

```python
print("📈 Quick Statistics:")
print(f"  Average Price: ₹{df['AskPrice'].mean():,.0f}")
print(f"  Price Range: ₹{df['AskPrice'].min():,.0f} - ₹{df['AskPrice'].max():,.0f}")
print(f"  Average Age: {df['Age'].mean():.1f} years")
print(f"  Average KM: {df['kmDriven'].mean():,.0f}")
print(f"  Most Common Brand: {df['Brand'].mode()[0]}")
print(f"  Most Common Fuel: {df['FuelType'].mode()[0]}")
```

```python
numerical_cols = df.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols = df.select_dtypes(include=['object', 'string']).columns.tolist()

print("Numerical Columns:", numerical_cols)
print("\nCategorical Columns:", categorical_cols)
```
